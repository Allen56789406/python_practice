```python
import pandas as pd
```

```python
import sqlite3
```

```python
conn=sqlite3.connect('text.db')
```

```python
url = 'https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv'
```

```python
df=pd.read_csv(url)
```

```python
df.head()
```

```python
df.to_sql('titanic',conn,if_exists='replace',index=False)
```

```python
pd.read_sql('SELECT *FROM titanic', conn)
```

```python
pd.read_sql('SELECT Name, Sex, Age, Survived FROM titanic', conn)
```

```python
pd.read_sql('SELECT * FROM titanic WHERE Survived = 1 ORDER BY AGE DESC', conn )
```

```python
result=pd.read_sql('SELECT Sex, Pclass, Survived FROM titanic', conn )
```

```python
result.pivot_table(values='Survived', index='Sex', columns='Pclass', aggfunc='mean')
```

```python
survived=pd.read_sql('SELECT * FROM titanic WHERE Survived =1', conn)
```

```python
survived.pivot_table(values='PassengerId',index='Sex', columns='Pclass', aggfunc='count')
```

```python
survived['AgeGroup']=pd.cut(survived['Age'], bins=[0, 18, 35, 60, 100], labels=['child', 'young','adult','senior'])
```

```python
survived.groupby('AgeGroup')['PassengerId'].count()
```
